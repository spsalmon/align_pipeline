# Custom Building Blocks

This block lets you run any alien script as part of the pipeline, and integrate its results into the filemap.

As it's impossible to predict what the output will be, the rerun option for this block is way less smart. If **rerun_custom** is False, the script won't run if the folder or file it's supposed to output already exists. If it is set to True, then everything is rerun.

It is important that the output of your script follows the general scheme of the pipeline. If your script outputs a folder of images, their names should contain the Time and Point information (as specified by **time_regex** and **point_regex** in the config) so that they can be properly linked to the rest of the filemap. If your script outputs a report file, it should contain "Time" and "Point" columns so that the results can be properly joined to the rest of the filemap.

## Options

* **custom_script_path** : The path to your custom script. It can either be a python '.py' or a bash '.sh' file. The file has to exist, this is checked before the run starts.
* **custom_script_name** : The name that will be used for naming the filemap column, and the output file / output folder.
* **custom_script_return_type** : "subdir" if your script outputs a directory of images, "csv" if it outputs a report file (kinda obvious), or null if your script outputs nothing.
  * With null, the block runs and hands over to the next one as usual, but nothing is added to the filemap. Use it when what your script produces is not per-timepoint and so has no row to attach to — one movie per point, a set of plots, a summary figure. Your script still receives `--output`, but it is passed as the string "None", so decide where to write things yourself (a parameter in **custom_script_parameters** is the usual way).
* **custom_script_parameters** : A list of parameters that will be added to the command when your script is called.
* **custom_script_requires_gpu** : Set to True if your script needs a GPU, so that the block is submitted with one. Defaults to **False**.

By default every custom script is fed the path to the analysis filemap through the `--filemap` argument, the output path through the `--output` argument, the block's configuration in pickled form through the `--block_config` argument, and the global configuration in pickled form through the `--config` argument, so your custom scripts should have a way to parse them.

Custom blocks work with both the `slurm` and the `local` backend.

### Example

```yaml
building_blocks:
  - "custom"

rerun_custom: [ False ]
custom_script_path: [ "./custom_workers/pumping_rate.py" ]
custom_script_name: [ "pumping_rate" ]
custom_script_return_type: [ "csv" ]
custom_script_requires_gpu: [ False ]
custom_script_parameters: [ [ "--input_column analysis_movies/raw_movies_str", "--frame_rate 25", "--height 1500", "--std_coeff 1", "--dist 3", "--width 1" ] ]
```

You will find an example of a script measuring the pumping rate in high frame rate movies in `custom_workers/pumping_rate.py`. You can use it as a template for your own custom scripts.
